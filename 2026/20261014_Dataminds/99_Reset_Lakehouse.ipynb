```python
LAKEHOUSE = "LH_Contoso"

tables = [r.tableName for r in spark.sql(f"SHOW TABLES IN {LAKEHOUSE}.dbo").collect()]
tables = [t for t in tables if not t.startswith("inventory")]
print("Te droppen:", tables)

for t in tables:
    spark.sql(f"DROP TABLE IF EXISTS {LAKEHOUSE}.dbo.{t}")
    print(f"Gedropt: {t}")
```
